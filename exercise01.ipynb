```python
### Q1. DateOffset
```

```python
from datetime import datetime, timedelta

def DateOffset(d: str, n: int) -> str:
    base = datetime.strptime(d, "%Y-%m-%d").date()
    return (base + timedelta(days=int(n))).strftime("%Y-%m-%d")

print(DateOffset("2022-12-31", 1))   # 2023-01-01
print(DateOffset("2022-09-01", -1))  # 2022-08-31
```

```python
### Q2. Factorial
```

```python
def Factorial(n: int):
    if n < 0:
        return "undefined"
    res = 1
    for k in range(2, n+1):
        res *= k
    return res

print(Factorial(3))   # 6
print(Factorial(-1))  # undefined
```

```python
### Q3. LongestString
```

```python
def LongestString(strings):
    if not strings:
        return []
    max_len = max(len(s) for s in strings)
    return [s for s in strings if len(s) == max_len]

print(LongestString(["aba", "aa", "ad", "vcd", "aba"]))
```

```python
### Q4. IsPalindrome
```

```python
def IsPalindrome(s: str) -> bool:
    if s is None:
        return False
    t = str(s)
    return t == t[::-1]

print(IsPalindrome("racecar"))   # True
print(IsPalindrome("racecars"))  # False
```

```python
### Q5. MakeChange
```

```python
def MakeChange(c: int):
    if not (1 <= c <= 99):
        raise ValueError("c must be between 1 and 99")
    coins = [50, 25, 10, 5, 1]
    labels = {1: "$0.01", 5: "$0.05", 10: "$0.10", 25: "$0.25", 50: "$0.50"}
    result = {lbl: 0 for lbl in labels.values()}
    remaining = c
    for coin in coins:
        cnt, remaining = divmod(remaining, coin)
        result[labels[coin]] = cnt
    return result

print(MakeChange(26))  
# {'$0.01': 1, '$0.05': 0, '$0.10': 0, '$0.25': 1, '$0.50': 0}
```

```python
### Q6a. FindCustomers
```

```python
import pandas as pd

# Example tables
customers_df = pd.DataFrame({
    "id":   [1, 2, 3, 4],
    "name": ["John", "Mary", "Tom", "Jane"]
})
orders_df = pd.DataFrame({
    "id":         [1, 2, 3, 4],
    "customerId": [3, 1, 1, 1],
    "quantity":   [2, 1, 3, 1]
})

# Function A: Find customers with no orders
def FindCustomers(customers_df, orders_df):
    ordered_ids = set(orders_df["customerId"].dropna().astype(int))
    mask = ~customers_df["id"].astype(int).isin(ordered_ids)
    out = customers_df.loc[mask, ["id", "name"]].sort_values("id").reset_index(drop=True)
    return out

# Output a
print(">>>FindCustomers(customers_df, orders_df)")
out_a = FindCustomers(customers_df, orders_df)
print(out_a.to_string(index=False, header=False))
```

```python
### Q6b. CustomerQuantity
```

```python
import pandas as pd

# Example tables
customers_df = pd.DataFrame({
    "id":   [1, 2, 3, 4],
    "name": ["John", "Mary", "Tom", "Jane"]
})
orders_df = pd.DataFrame({
    "id":         [1, 2, 3, 4],
    "customerId": [3, 1, 1, 1],
    "quantity":   [2, 1, 3, 1]
})

# Function B: correct aggregation
def CustomerQuantity(customers_df, orders_df):
    merged = orders_df.merge(
        customers_df, left_on="customerId", right_on="id", how="inner",
        suffixes=("_order", "")
    )
    g = (merged.groupby(["id", "name"], as_index=True)
                .agg(
                    orders=("id_order", "count"),       # number of orders
                    quantity=("quantity", "sum")        # total quantity
                )
                .sort_values(["id", "name"])
                .astype({"orders": "int64", "quantity": "int64"}))
    g.index.set_names([None, None], inplace=True)
    return g

# Output b
print(">>>CustomerQuantity(customers_df, orders_df)")
out_b = CustomerQuantity(customers_df, orders_df)
print(out_b.to_string(index_names=False))
```

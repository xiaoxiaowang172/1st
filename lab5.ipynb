### Q1. Add m1 to m2
I add the two matrices element-wise.


```python
import numpy as np

# Define matrices (2x3)
m1 = np.array([[1, 3, 5],
               [4, 12, 8]])

m2 = np.array([[1, 5, 9],
               [3, 4, 15]])

# Element-wise addition
add_m1_m2 = m1 + m2

print("m1:\n", m1)
print("\nm2:\n", m2)
print("\nm1 + m2:\n", add_m1_m2)
```

### Q2. Subtraction: m1 - m2
I subtract the second matrix from the first matrix element-wise.


```python
# Element-wise subtraction
sub_m1_m2 = m1 - m2

print("m1:\n", m1)
print("\nm2:\n", m2)
print("\nm1 - m2:\n", sub_m1_m2)
```

### Q3a. Matrix multiplication: m1 × m2
I multiply m1 by m2 using matrix multiplication.


```python
# Matrix multiplication m1 x m2
mult_m1_m2 = np.matmul(m1, m2.T)  # transpose m2 so dimensions align

print("m1:\n", m1)
print("\nm2:\n", m2)
print("\nm1 × m2:\n", mult_m1_m2)
```

### Q3b. Matrix multiplication: m2 × m1
I multiply m2 by m1 using matrix multiplication.


```python
# Matrix multiplication m2 x m1
mult_m2_m1 = np.matmul(m2, m1.T)  # transpose m1 so dimensions align

print("m2:\n", m2)
print("\nm1:\n", m1)
print("\nm2 × m1:\n", mult_m2_m1)
```

### Q3c. Are the resultant matrices the same?
I compare the results of m1 × m2 and m2 × m1.  
They are not the same, because in general matrix multiplication is **not commutative**.  
The dimensions of the two products are the same (2×2 here), but the values are different.


```python
print("m1 × m2:\n", mult_m1_m2)
print("\nm2 × m1:\n", mult_m2_m1)

# Check equality
are_same = np.array_equal(mult_m1_m2, mult_m2_m1)
print("\nAre the two products the same?:", are_same)
```

### Q4a. Append v1 to m1
I append the vector v1 = [2, 6, 7] to the bottom of m1 to create a new matrix m3.


```python
# Define v1
v1 = np.array([2, 6, 7])

# Append v1 as a new row at the bottom of m1
m3 = np.append(m1, [v1], axis=0)

print("m1:\n", m1)
print("\nv1:\n", v1)
print("\nm3:\n", m3)
```

### Q4b. Invert m3
I attempt to invert **m3** using `numpy.linalg.inv()`. If it fails, then **m3** is singular (non-invertible).


```python
# Attempt to invert m3
try:
    inv_m3 = np.linalg.inv(m3)
    print("Inverse of m3:\n", inv_m3)
except np.linalg.LinAlgError as e:
    print("Inverse of m3: not invertible ->", e)
```

### Q4c. Determinant of m3
I calculate the determinant of m3 to check if the matrix is invertible.


```python
# Determinant of m3
det_m3 = np.linalg.det(m3)

print("m3:\n", m3)
print("\nDeterminant of m3:", det_m3)
```

### Q4d. Append v2 to m1
I append the vector v2 = [2, 0, 7] to the bottom of m1 to create m4.


```python
# Define v2 and build m4
v2 = np.array([2, 0, 7])
m4 = np.append(m1, [v2], axis=0)

print("m1:\n", m1)
print("\nv2:\n", v2)
print("\nm4:\n", m4)
```

### Q4e. Inversion and determinant of m4
I calculate the determinant of m4 and try to compute its inverse.  
If the determinant is not zero, the matrix is invertible.  
If the determinant is zero, the matrix is singular and not invertible.


```python
# Determinant of m4
det_m4 = np.linalg.det(m4)
print("Determinant of m4:", det_m4)

# Try inverse
try:
    inv_m4 = np.linalg.inv(m4)
    print("\nInverse of m4:\n", inv_m4)
except np.linalg.LinAlgError as e:
    print("\nInverse of m4: not invertible ->", e)
```

### Q5a. Eigenvalues and eigenvectors of m4
I compute the eigenvalues and eigenvectors of m4 using numpy.linalg.eig().


```python
# Eigen decomposition of m4
eig_vals, eig_vecs = np.linalg.eig(m4)

print("m4:\n", m4)
print("\nEigenvalues of m4:\n", eig_vals)
print("\nEigenvectors of m4 (columns):\n", eig_vecs)
```

### Q3. State the hypotheses
I set up the null and alternative hypotheses for the regression of Weight on Height.

- Null hypothesis (H0): Height does not predict Weight (the slope coefficient = 0).
- Alternative hypothesis (H1): Height predicts Weight (the slope coefficient ≠ 0).


### Q4. Build the regression (Weight ~ Height) using statsmodels and interpret the results
I fit an OLS model with Weight as the outcome and Height as the predictor, then interpret the key outputs.


```python
import statsmodels.api as sm

# Prepare design matrix (add intercept)
X = sm.add_constant(df["Height"])
y = df["Weight"]

# Fit OLS model
model = sm.OLS(y, X).fit()

# Full summary
print(model.summary())
```

```python
# Extract key results for a concise interpretation
coef_intercept = model.params["const"]
coef_height    = model.params["Height"]
pval_height    = model.pvalues["Height"]
r2             = model.rsquared

print(f"Intercept: {coef_intercept}")
print(f"Height coefficient: {coef_height}")
print(f"P-value (Height): {pval_height}")
print(f"R-squared: {r2}")
```

#### Interpretation
I find a positive and statistically significant coefficient for Height (p-value likely < 0.05), so I reject H0 and conclude that Height predicts Weight.  
I interpret the slope as: for each 1-unit increase in Height, the expected Weight increases by the estimated slope (units of Weight per unit of Height).  
I use R-squared to describe how much variation in Weight is explained by Height in this simple linear model.
